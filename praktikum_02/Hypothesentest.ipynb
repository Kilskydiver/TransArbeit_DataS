```python
from scipy import stats

gruppe_a = [23, 25, 21, 27, 24, 26]
gruppe_b = [30, 28, 32, 29, 31, 33]

t_stat, p_wert = stats.ttest_ind(gruppe_a, gruppe_b)

if p_wert < 0.05:
    print("H0-verwerfen: Es gibt einen signifikanten Unterschied zwischen den beiden Gruppen.")
else:
    print("Es gibt keinen signifikanten Unterschied zwischen den beiden Gruppen.")
```
