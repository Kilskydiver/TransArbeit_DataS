```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('shoporders.csv',
                 sep=',',
                 encoding='utf-8',
                 decimal=',',
                 parse_dates=['order_date']
                 )

df
```

```python


df['shipping_country'].value_counts(dropna=False)
```

```python
# Normalisiere shipping_country auf genau diese Zielwerte:
# Schweiz, Italien, Frankreich, Deutschland, Österreich

country_map = {
    # Schweiz
    "schweiz": "Schweiz",
    "switzerlaand": "Schweiz",
    "switerland": "Schweiz",
    "switzerland": "Schweiz",
    "ch": "Schweiz",

    # Italien
    "italien": "Italien",
    "italy": "Italien",

    # Deutschland
    "deutschland": "Deutschland",
    "deuschland": "Deutschland",
    "germany": "Deutschland",

    # Frankreich
    "frankreich": "Frankreich",
    "france": "Frankreich",

    # Österreich
    "österreich": "Österreich",
    "osterreich": "Österreich",
    "oestereich": "Österreich",
    "ostereich": "Österreich",
    "austria": "Österreich",
}

df["shipping_country"] = (
    df["shipping_country"]
    .astype("string")
    .str.strip() 
    .str.lower()
    .map(country_map)
)

df
```

```python
# Doppelte Zeilen entfernen und das DataFrame aktualisieren
df = df.drop_duplicates()
df
```

```python
# Wechselkurse (Beispielwerte; anpassen, falls du andere Kurse brauchst)
fx_rate = {
    "CHF": 1.0,
    "EUR": 0.96,   # 1 EUR = 0.96 CHF
    "USD": 0.88,   # 1 USD = 0.88 CHF
}
df['price'] = df['price'].fillna(0)

# Normalisiere Währungen
df["currency"] = df["currency"].astype("string").str.upper().str.strip()

# Preis in CHF umrechnen
df["price_chf"] = (
    df["price"]
    .astype(float)
    * df["currency"].map(fx_rate).fillna(1.0)
).round(2)

df
```

```python
# Gesamtpreis berechnen; ein Dutzend entspricht 12 Stück
einheiten_pro_position = (
    df["unit"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map({
        "stück": 1,
        "stueck": 1,
        "stück": 1,
        "dutzend": 12,
        "dozen": 12,
    })
    .fillna(1)
)

df['price'] = df['price'].fillna(0)

df["preis_total"] = (
    df["price_chf"].astype("Float64")
    * df["quantity"].astype("Float64")
    * einheiten_pro_position
).round(2)

df

```

```python
df_ohe = pd.get_dummies(df, columns=["payment_method"], prefix="pay")

df_ohe
```

```python
df_ohe["weekday"] = df_ohe['order_date'].dt.dayofweek

df_ohe["month"] = df_ohe['order_date'].dt.month

df_ohe["is_weekend"] = df_ohe['weekday'] >=5

df_ohe
```

```python
from sklearn.preprocessing import(
    StandardScaler,
    MinMaxScaler
)

scaler = StandardScaler()
df_ohe["price_scaled"] = scaler.fit_transform(df_ohe[["price_chf"]])

df_ohe

```
